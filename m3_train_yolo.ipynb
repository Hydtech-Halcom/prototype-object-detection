```
# 3. Download the YOLO dataset exported from Roboflow (copy this code from Roboflow > Export)
from roboflow import Roboflow
rf = Roboflow(api_key="YOUR_API_KEY")
dataset = rf.workspace("productionai").project("object-detection").version(1).download("yolov11")
```

```
# 4. Train
from ultralytics import YOLO
model = YOLO("yolo11n.pt")
model.train(data=f"{dataset.location}/data.yaml", epochs=100, imgsz=640)
```

```
# 5. Evaluate (target: mAP50 above 0.90)
metrics = model.val()
print("mAP50:", metrics.box.map50, " mAP50-95:", metrics.box.map)
print("Precision:", metrics.box.mp, " Recall:", metrics.box.mr)
```

```
# 6. Download the trained model
from google.colab import files
files.download(f"{model.trainer.save_dir}/weights/best.pt")
```
